```python
def search_by_verb(verb):
    return train_df[train_df["verb"].str.lower() == verb.lower()]
```

```python
search_by_verb("cut").head()
```

```python
def search_narration(keyword):
    return train_df[
        train_df["narration"].str.contains(
            keyword,
            case=False,
            na=False
        )
    ]
```

```python
search_narration("onion")
```
